## Question 4 (optional): Stability of a system by linearisation

$$
\dot z=f(z),\qquad z\in\mathbb R^2,
$$

$$
\dot{\delta z}=J(z^\star)\delta z,
$$

$$
J(z)=
\begin{bmatrix}
\partial f_1/\partial y & \partial f_1/\partial x\\
